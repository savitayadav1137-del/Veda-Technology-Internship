```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("C:\Customer_data.csv")
print(df)
df.head()
```

```python
df.info()
print(df.isnull().sum())
```

```python
text_column=df.select_dtypes(include=["object","string"]).columns.to_list()
print(text_column)
```

```python
cleaned_df=df.copy()
```

```python
import re

cleaned_df = df.copy()

text_columns = cleaned_df.select_dtypes(
    include="object"
).columns

for col in text_columns:
    cleaned_df[col] = cleaned_df[col].apply(
        lambda x: re.sub(r"\s+", " ", x.strip())
        if isinstance(x, str) else x
    )

print("Text cleaning completed!")
```

```python
for col in cleaned_df.columns:
    if "name" in col.lower():
        cleaned_df[col] = cleaned_df[col].str.title()

    elif "email" in col.lower():
        cleaned_df[col] = cleaned_df[col].str.lower()

print("Names and emails cleaned!")
```

```python
print("Before Cleaning:")
display(df.head())

print("After Cleaning:")
display(cleaned_df.head())
```

```python
cleaned_df.to_csv("cleaned_customer_data.csv", index=False)

print("Cleaned dataset saved!")
```

```python
comparison = pd.DataFrame({
    "before": df.select_dtypes(include="object").iloc[:, 0],
    "after": cleaned_df.select_dtypes(include="object").iloc[:, 0]
})

comparison.to_csv("before_after_comparison.csv", index=False)

print("Comparison file saved!")
```
