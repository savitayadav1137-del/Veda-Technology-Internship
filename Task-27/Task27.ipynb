```python
df = pd.read_csv('retail_sales_dataset.csv')

df.head()
```

```python
df['Date'].head(10)
```

```python
print(df['Date'].dtype)
```

```python
df['Date'] = pd.to_datetime(
    df['Date'],
    format='%Y-%m-%d',
    errors='coerce'
)

df.head()
```

```python
print(df['Date'].dtype)
```

```python
invalid_dates = df[df['Date'].isna()]

print("Invalid dates:", len(invalid_dates))

invalid_dates
```

```python
issue_log = df[df['Date'].isna()].copy()

issue_log['Issue'] = 'Invalid date'

issue_log
```

```python
df['Date'].dt.strftime('%d-%m-%Y').head(10)
```

```python
print("Total rows:", len(df))
print("Missing dates:", df['Date'].isna().sum())
print("Date datatype:", df['Date'].dtype)

df.head()
```

```python
df.to_csv('cleaned_retail_sales.csv', index=False)

issue_log.to_csv('date_issue_log.csv', index=False)

print("Both files saved successfully!")
```
