**Dataset Upload and read**

```python
import pandas as pd
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
display(df.head())
```

**DataSet**

```python
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

**Check Duplicate Records**

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

**Display Duplicate Record**

```python
duplicates = df[df.duplicated(keep=False)]

print("Duplicate Records:")
display(duplicates)
```

**Check Duplicate in Key Column**

```python
duplicate_ids = df[
    df.duplicated(subset=["Transaction ID"], keep=False)
]

print("Duplicate Transaction IDs:", len(duplicate_ids))
display(duplicate_ids)
```

**Check Missing Values**

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

**Duplicate Report**

```python
duplicate_report = duplicates.copy()

duplicate_report.to_csv(
    "Duplicate_Report.csv",
    index=False
)

print("Duplicate report saved successfully!")
```

**Cleaned Dataset**

```python
df_cleaned = df.drop_duplicates().copy()

print("Original Rows:", len(df))
print("Rows After Cleaning:", len(df_cleaned))
print("Duplicates Removed:", len(df) - len(df_cleaned))
```

```python
print(
    "Duplicates Remaining:",
    df_cleaned.duplicated().sum()
)
```

**Audit Report**

```python
audit_report = pd.DataFrame({
    "Metric": [
        "Original Total Rows",
        "Original Total Columns",
        "Duplicate Rows Found",
        "Duplicate Transaction IDs",
        "Rows After Cleaning",
        "Duplicates Remaining",
        "Verification Status"
    ],
    "Result": [
        len(df),
        df.shape[1],
        duplicate_count,
        len(duplicate_ids),
        len(df_cleaned),
        df_cleaned.duplicated().sum(),
        "Verified"
    ]
})

display(audit_report)

audit_report.to_csv(
    "Duplicate_Audit_Report.csv",
    index=False
)
```

**Save Cleaned Copy**

```python
df_cleaned.to_csv(
    "Retail_Sales_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
